# **Business Decision Making**
The decision-making process in business is fundamental to the strategic functioning of organizations and occurs at all hierarchical levels, from the operational base to senior management.
```python
# Importing all the necessary libraries and resources:
import numpy as np
import pandas as pd
from scipy.optimize import linprog
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

## **Example: Quantitative Methods**
Quantitative methods are based on numerical data and mathematical models and are widely used for decisions involving measurable factors such as cost, time, and resources. Multi-criteria decision analysis methods, such as AHP, Electre, and Promethee, allow alternatives to be evaluated objectively based on multiple criteria. Mathematical modeling and simulation techniques, including linear optimization, Monte Carlo simulation, and sensitivity analysis, help predict outcomes and assess the impact of different decisions. Additionally, machine learning algorithms—such as linear regression, decision trees, and neural networks—analyze large volumes of data to identify trends and support decision-making, often implemented using tools like Python with libraries such as NumPy and Pandas.

```python
# Quantitative Methods Example: Linear Optimization + Linear Regression.
# Converting to minimization (negative profit):
c = [-40, -30] # Corrected numerical values
A = [[2, 3], [1, 1]] # Corrected matrix A (assuming common linear programming problem structure)
b = [10, 8] # Corrected vector b

# Corrected linprog function call and variable assignment
result = linprog(c, A_ub=A, b_ub=b, bounds=(0, None)) # Corrected function call and 'No'e' to 'None'

# Corrected print statements
print('Optimal Production Plan:') # Corrected string literal
print(f'Product X: {result.x[0]:.2f}') # Corrected f-string and print call
print(f'Product Y: {result.x[1]:.2f}') # Corrected f-string and print call
print(f'Maximum Profit: ${-result.fun:.2f}') # Corrected f-string (removed extra quote)

# Machine Learning
# Predict sales based on advertising budget
data = pd.DataFrame({
    'Advertising': [10, 20, 30, 40, 50],
    'Sales': [15, 25, 35, 45, 60]
})

X = data[['Advertising']]
y = data['Sales']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = LinearRegression()
model.fit(X_train, y_train)

# Corrected prediction input to be a DataFrame with feature names
prediction = model.predict(pd.DataFrame({'Advertising': [60]}))

print("\nPredicted Sales for $60 Advertising:", prediction[0])
```

## **Example: Qualitative Methods**
Qualitative methods are more subjective and rely on judgments and opinions, being used when decision factors cannot be easily quantified. Tools such as SWOT analysis, scenario analysis, and cognitive mapping help decision-makers evaluate internal and external factors, explore possible future situations, and structure complex thinking. In decision theory, criteria like cost, quality, and time are used to compare alternatives, often through multi-criteria approaches that enable more consistent and robust analyses. Data normalization is essential to ensure models correctly interpret information, especially when minimizing or maximizing specific criteria. By combining structured evaluation, risk analysis, mathematical techniques, and machine learning tools—implemented with software such as R, Matlab, and Python—decision-makers can better predict outcomes and minimize negative consequences.

```python
# Qualitative Methods Example: SWOT + Multi-Criteria Decision Analysis.
# SWOT Analysis:
swot = {
    'Strengths': ['Strong brand', 'Skilled team'],
    'Weaknesses': ['High costs', 'Limited capacity'],
    'Opportunities': ['Growing market', 'New technology'],
    'Threats': ['Competition', 'Regulation']
}

print('SWOT Analysis:')
for category, items in swot.items():
    print(f'{category}: {', '.join(items)}')
```

```python
# Multi-Criteria Decision.
# Alternatives: A, B, C; Criteria: Cost (minimize), Quality (maximize), Time (minimize)
data = pd.DataFrame({
    'Alternative': ['A', 'B', 'C'],
    'Cost': [100, 120, 90],
    'Quality': [80, 85, 70],
    'Time': [30, 25, 35]
})

# Normalizing data:
data['Cost_norm'] = (data['Cost'].max() - data['Cost']) / (data['Cost'].max() - data['Cost'].min())
data['Quality_norm'] = (data['Quality'] - data['Quality'].min()) / (data['Quality'].max() - data['Quality'].min())
data['Time_norm'] = (data['Time'].max() - data['Time']) / (data['Time'].max() - data['Time'].min())

# Weights:
weights = {
    'Cost_norm': 0.4,
    'Quality_norm': 0.4,
    'Time_norm': 0.2
}

# Calculating weighted score:
data['Score'] = (
    data['Cost_norm'] * weights['Cost_norm'] +
    data['Quality_norm'] * weights['Quality_norm'] +
    data['Time_norm'] * weights['Time_norm']
)

print('\nDecision Scores:')
print(data[['Alternative', 'Score']])

best = data.loc[data['Score'].idxmax()]
print(f'\nBest Alternative: {best['Alternative']}')
```
