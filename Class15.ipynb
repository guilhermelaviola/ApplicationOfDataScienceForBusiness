# **Tokenization & Count Vectorization**
The text explains that tokenization is a fundamental natural language processing (NLP) technique that transforms unstructured text into structured data by dividing it into smaller units such as words, sub-words, characters, or sentences. This process enables machine learning algorithms to analyze text effectively for business applications like sentiment analysis, document classification, clustering, topic analysis, and feature extraction. Different tokenization methods—word, sub-word, character, sentence, and regular expression tokenization—serve distinct analytical purposes depending on the objective and context. Complementing tokenization, vectorization converts tokens into numerical representations using techniques such as Count Vectorization (Bag-of-Words), TF-IDF, word embeddings, and contextual embeddings, allowing algorithms to process textual data mathematically. In business environments, these techniques support strategic decision-making by extracting insights from customer feedback, documents, and other textual sources, typically through structured processes like ETL (Extract, Transform, Load) and supported by appropriate IT infrastructure. Overall, tokenization and vectorization are essential tools for converting raw textual data into actionable business intelligence.


```python
# Importing all the necessary libraries:
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from sklearn.feature_extraction.text import CountVectorizer
nltk.download('punkt')
nltk.download('punkt_tab') # Download punkt_tab
```

## **Example: Tokenization in Business**
Tokenization is a powerful tool not only for sentiment analysis, but also for other applications, such as creating word clouds and word trees. By counting the frequency of words in a text (Count Vectorization), we can visualize which terms are most common and identify patterns that are important for the business.

```python
# Defining a text for analysis and applying word and sentence tokenization:
text = 'Tokenization is a fundamental step in natural language processing. It is essential for sentiment analysis.'

tokenized_words = word_tokenize(text)
tokenized_sentences = sent_tokenize(text)

print('Original text', text)
print('Tokenized words', tokenized_words)
print('Tokenized senteces', tokenized_sentences)
```

```python
# Defining a list of documents (texts) and transforming them into numeric vectors based on word frequency:
documents = [
    'Tokenization is a fundamental step in natural language processing.',
    'It is essential for sentiment analysis and other data processing.'
]

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(documents)

print('Vocabulary words:', vectorizer.get_feature_names_out())
print('Vector representation of documents:\n', X.toarray())
```
